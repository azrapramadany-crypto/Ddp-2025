```python
# celciuskefahreinheit

def celcius_ke_fahrenheit(c):
    return (c * 9/5) + 32

print(celcius_ke_fahrenheit(0))    # Output: 32.0
print(celcius_ke_fahrenheit(100))  # Output: 212.0
```

```python
# fungsi bernama is genap dan false
def is_genap(angka):
    return angka % 2 == 0

print(is_genap(4))  # Output: True
print(is_genap(7))  # Output: False
```

```python
# fungsi untuk melihat keterangan lulus atau tidak lulus

def nilai(n):
    if n >= 70:
        return "lulus"
    else:
        return "gagal"

print(nilai(80))  # lulus
print(nilai(60))  # gagal
```

```python
# fungsi untuk menampilkan bilangan ganjil

def bilangan(n):
    ganjil = []
    for i in range(1, n):
        if i % 2 != 0:
            ganjil.append(i)
    return ganjil

print(bilangan(20))
```
